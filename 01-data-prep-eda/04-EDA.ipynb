#### Grouping Data

```python
groceries.head(2)
```

```python
groceries.groupby('Category')['Inventory'].sum() # group by categories looking at total inventory within
```

```python
groceries.groupby('Category')['Inventory'].count()
```

```python
groceries.groupby(['Category', 'Subcategory'])['Inventory'].count().reset_index().sort_values('Inventory') # group by 2 columns
```

```python
groceries.groupby(['Category', 'Subcategory'])['Inventory'].agg(['sum', 'count']).reset_index() # see several aggregation columns
```

```python
# putting in brackets long chained code
(
    groceries
    .groupby(['Category', 'Subcategory'])['Inventory']
    .agg(['sum', 'count'])
    .reset_index()
)
```

```python
groceries.groupby('Category')[['Item', 'Price_Dollars']].max() # it is incorrect! Item is not going to be correct associated with the price item!!!
# max() applies to both columns separately!
```

```python
# how to find the most expensive item in each category - CORRECT approach
# 1 - narrow down data
# 2 - sort it, so that the most expensive would be at the top
# 3 - then when I group by category and look at first row in the group - it's goint to lead to the most expensive item
(
    groceries[['Category', 'Item', 'Price_Dollars']]
    .sort_values('Price_Dollars', ascending=False)
    .groupby('Category')
    .head(1)
)   
```

```python
happiness = pd.read_csv('data/happiness_survey_data.csv')
```
